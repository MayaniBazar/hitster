```python
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials

# 1. Your Spotify developer credentials
client_id = "eb79357f6750466694c965d8487a65bc"
client_secret = "XXX"

# 2. Setup the authorization manager
auth_manager = SpotifyClientCredentials(client_id=client_id, client_secret=client_secret)
sp = spotipy.Spotify(auth_manager=auth_manager)

# 3. Provide the playlist link or URI
playlist_url = "https://open.spotify.com/playlist/1JxO5N9LoXvIduJzMQWc6Q?si=e2ea0cedc7604849"  # Example
playlist_id = playlist_url.split("/")[-1].split("?")[0]

# 4. Fetch the playlist tracks
results = sp.playlist_tracks(playlist_id)

# 5. Extract song data
songs = []
for item in results['items']:
    track = item['track']
    name = track['name']
    artists = ", ".join([artist['name'] for artist in track['artists']])
    release_date = sp.track(track['id'])['album']['release_date']
    year = release_date.split("-")[0]

    songs.append({
        'name': name,
        'artist': artists,
        'year': year
    })

# 6. Print or export
for song in songs:
    print(f"{song['name']} - {song['artist']} ({song['year']})")
```
